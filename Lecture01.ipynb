```python
# Pipeline Concept : Orchestrator(Architecture)Monotonous code

def sum(num1,num2): # function definition : sum(num1,num2)
    return(num1 + num2)

def sub(num1,num2):
    return(num1 - num2)

def mul(num1,num2):
    return(num1 * num2)

def div(num1,num2):
    return(num1 / num2)

def main():   # Function Definition

    num1 = float(input("Enter first number :"))
    num2 = float(input("Enter second number :"))

    choice = input("Enter the choice ")

    if choice == 'sum':
        # Addition
        ans = sum(num1,num2) # Arguments
    elif choice =='sub':
        # Subtraction
        ans = sub(num1,num2) # function calling : sum(num1,num2)
    elif choice == 'mul':
        # Multiplication
        ans = mul(num1,num2)
    elif choice == 'div':
        # Division
        ans= div(num1,num2)
    else:
        print("Invalid Choice")
    print(ans)

main()        # Function calling : Orcehstrator
```

```python
 #Inheritence and its Types.
 # Single: Child inherits from single parent class.
 # Hybrid: Combination of two or more types of inheritance.
 # Hierarchical: Multiple child inherit from single parent class.
 # Multilevel: Classes inherit in a chain from parent to child.

class employee:
    # Define constructor
    def __init__(self,name,age,
                 department,salary):
        self.name = name
        self.age = age
        self.department = department
        self.salary = salary

    def employee_info(self):
        print("Name:", self.name)
        print("Age:", self.age)
        print("Department:", self.department)
        print("Salary:", self.salary)
        
class manager(employee):
        
    # Define Constructor
    def __init__(self,name,age,
                 department,salary,designation):
        super().__init__(name,age,salary,department)   # Super() funtion is calling a constructor with certain arguments by using . operator
        self.designation = designation

    def info(self):
        print(f'The Name of Manager is {self.name}')
        print(f'The Designation is {self.designation}')

# Define Object
emp1 = employee('Darshan',21,'HR',200000)
emp2 = manager('Riya',21,'IT',300000,'Admin')
```

```python
# WAP Hierachical structure for banking application/ Train system Pipeline

# Parent Class
class ticket:
    def __init__(self,ticket_id,passenger_name,train_no, fare):
        self.ticket_id = ticket_id
        self.passenger_name = passenger_name
        self.train_no = train_no
        self.fare = fare

    def generate_ticket(self):
        print("Ticket generated successfully.")

    def cancel_ticket(self):
        print("Ticket cancelled successfully.")

    def print_receipt(self):
        print("\n----- Ticket Receipt -----")
        print("Ticket ID  :", self.ticket_id)
        print("Passenger Name :", self.passenger_name)
        print("Train Number  :", self.train_no)
        print("Fare :", self.fare)

# Child Class 1
class sleeper_ticket(ticket):
    def __init__(self, ticket_id, passenger_name, train_no, fare, berth_no):
        super().__init__(ticket_id, passenger_name, train_no, fare)
        self.berth_no = berth_no

# Child Class 2
class ac_ticket(ticket):
    def __init__(self, ticket_id, passenger_name, train_no, fare, coach):
        super().__init__(ticket_id, passenger_name, train_no, fare)
        self.coach = coach

# Child Class 3
class general_ticket(ticket):
    def __init__(self, ticket_id, passenger_name, train_no, fare):
        super().__init__(ticket_id, passenger_name, train_no, fare)

# Creating objects
sleeper_1 = sleeper_ticket(101, "Aman", 12345, 750, "S12-45")
ac_1= ac_ticket(102, "Diya", 12345, 1800, "A2")
general_1 = general_ticket(103, "Ajay", 12345, 250)

```

```python
# Abstarct Classes without code implementation

from  abc import ABC,abstractmethod

class bank(ABC):

    @abstractmethod
    def credit_info(self):
        pass

    @abstractmethod
    def debit_info(self):
        pass

    @abstractmethod
    def balance_info(self):
        pass 
```

```python
class saving_acct(bank):

    def credit_info(self):
        print('Credit Info')

    def debit_info(self):
        print('Debit Info')

    def balance_info(self):
        print('Balance Info')
```

```python
# try except else block

num1 = 100
num2 = 20

try:
    result = num1 / num2

except ZeroDivisionError:
    print("Cannot divide by zero.")

else:
    print("Division Successful.")
    print("Result:", result)
```
