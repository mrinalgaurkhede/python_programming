```python
#TASK1
num1 = float(input("Enter the first number: "))
num2 = float(input("Enter the second number: "))

addition = num1 + num2
subtraction = num1 - num2
multiplication = num1 * num2
modulo = num1 % num2

print("Addition:", addition)
print("Subtraction:", subtraction)
print("Multiplication:", multiplication)
print("Modulo:", modulo)
```

```python
## Concept of Indexing and Slicing
'''
Indexing will give the idea of the location
Slicing means cutting the code in two parts
'''
a = "Hello World"
print(a[0::2])  #a[Start:Stop:Step]
print(a[-1::-1])
```

```python
# Write a program wherther the entered string is palindorm or not

string = input("Enter the string")

string1 = string[::-1] # reverse the given string

if string == string1:
    print("The given string is palindrom")
else:
    print('The given string is not a palindrom')
```
