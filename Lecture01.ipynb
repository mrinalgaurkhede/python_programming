```python
number = int(input("Enter the number"))      # number outside the function

def factorial(number):               # the whole thing is called argument
    factorial = 1                    # . 
    for i in range(1,number+1):      # .
        factorial = factorial * i    # .
    return factorial                 # .

ans = factorial(number)
print(ans)
```

```python
# Pipeline Concept : Orchestrator(Architecture)Monotonous code

def sum(num1,num2): # function definition : sum(num1,num2)
    return(num1 + num2)

def sub(num1,num2):
    return(num1 - num2)

def mul(num1,num2):
    return(num1 * num2)

def div(num1,num2):
    return(num1 / num2)

def main():   # Function Definition

    num1 = float(input("Enter first number :"))
    num2 = float(input("Enter second number :"))

    choice = input("Enter the choice ")

    if choice == 'sum':
        # Addition
        ans = sum(num1,num2) # Arguments
    elif choice =='sub':
        # Subtraction
        ans = sub(num1,num2) # function calling : sum(num1,num2)
    elif choice == 'mul':
        # Multiplication
        ans = mul(num1,num2)
    elif choice == 'div':
        # Division
        ans= div(num1,num2)
    else:
        print("Invalid Choice")
    print(ans)

main()        # Function calling : Orcehstrator
```

```python
''' 

Class and Object:
Class: blueprint or template/framework
Object: instance of a class

Note:
1. any function when defined inside a class is called "Methods of Respective class"
2. any variable when defined inside a class is called "Attribute of Respective class"

'''

# blueprint/template : class b
class bank:

# Constructor/Special Function/ Special Method
    def __init__(self,name):                  # Attribute1
        self.name = name

    def emp(self):                           # Method
        name = self.name                     # Attribute1

# Define object : sbi
sbi = bank('india')
```
