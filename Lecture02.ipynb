```python
# Create Custom Loggings: It is the process of recording tracking data and events while an application runs. 
import logging
logging.basicConfig(level = logging.INFO,
                    filename = 'railway.log',
                    filemode = 'w',
                    format = '%(asctime)s - %(levelname)s - %(message)s',
                    force = True)
```

```python
logging.info('Railone Application Started')
```

```python
logging.info('Railone Application Terminated')
```
