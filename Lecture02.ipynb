```python
#  Web Scrapping : the automated process of extracting large amounts of data from websites
import re

sentence = "Hello World"
print(sentence)
```

```python
sentence = r"\t\t\t\nHello World"
print(sentence)
```

```python
text_to_search = '''Hello World,I am excited to learn python programming'''
pattern = re.compile(r'a')
matches = pattern.finditer(text_to_search)

for i in matches:
    print(i)
```

```python
import re 
text = '''
Customer Name: Rahul Sharma
I am 25 years old
The phone number is 9876543210
Order Id:  ORD12345
Delivery Expected in 3 days
'''

result = re.findall("[0-9]",text)
print(result)
```
