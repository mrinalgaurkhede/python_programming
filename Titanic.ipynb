```python
from collections import OrderedDict
numerical_col = data.select_dtypes(exclude = 'object').columns
categorical_col = data.select_dtypes(include = 'object').columns
categorical_col

def stats(): # Function definition
  stats=[]
  for i in numerical_col:
    numerical_stats = OrderedDict({
      'Column_Name': i,
      'Maximum':data[i].max(),
      'Minimum':data[i].min(),
      'Mean':data[i].mean(),
      'Median':data[i].median(),
      'Q1':data[i].quantile(0.25),
      'Q3':data[i].quantile(0.75),
      'IQR':data[i].quantile(0.75) - data[i].quantile(0.25),
      'Variance':data[i].var(),
      'Standard Deviation':data[i].std(),
      'Skewness':data[i].skew(),
      'Kurtosis':data[i].kurt()
    })
    stats.append(numerical_stats)
    report = pd.DataFrame(stats)

  print(report)

stats()
```

```python
import pandas as pd

df = pd.read_csv("titanic-Dataset.csv")

print(df.head())
```

```python
# checking Descriptive stats

numerical_col = df.select_dtypes(exclude = 'object').columns
categorical_col = df.select_dtypes(include = 'object').columns

from collections import OrderedDict
stats = []
for i in numerical_col:
  numerical_stats = OrderedDict({
      'Feature': i,
      'Maximum':df[i].max(),
      'Minimum':df[i].min(),
      'Mean':df[i].mean(),
      'Median':df[i].median(),
      'Unique_Values':df[i].nunique(),
      '% of Missing Values':df[i].isnull().sum()/df.shape[0]*100,
      'Null Values':df[i].isnull().sum(),
      'Q1':df[i].quantile(0.25),
      'Q3':df[i].quantile(0.75),
      'IQR':df[i].quantile(0.75) - df[i].quantile(0.25),
      'Standard Deviation':df[i].std(),
      'Skewness':df[i].skew(),
      'Kurtosis':df[i].kurt()
  })
  stats.append(numerical_stats)
  report = pd.DataFrame(stats)

report
```

```python
report.to_csv('Report.csv')
```
